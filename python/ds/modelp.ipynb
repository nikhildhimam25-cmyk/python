```python
import pandas as pd 
```

```python
df=pd.read_csv("Used Car Dataset.csv")
```

```python
df.info()
df.head()
```

```python
df.drop("Unnamed: 0",axis=1,inplace=True)
df.drop("torque(Nm)",axis=1,inplace=True)
df.drop("max_power(bhp)",axis=1,inplace=True)
df.drop("registration_year",axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
enc=LabelEncoder()
hot=OneHotEncoder()
df
```

```python
fuel=enc.fit_transform(df['fuel_type'])
df.loc[:,["fuel_type"]]=fuel
```

```python
own=enc.fit_transform(df.loc[:,["ownsership"]])
df.loc[:,["ownsership"]]=own
```

```python
trans=enc.fit_transform(df.loc[:,["transmission"]])
df.loc[:,["transmission"]]=trans
```

```python
car=hot.fit_transform(df.loc[:,["car_name"]]).toarray()
df.loc[:,["car_name"]]=car
```
